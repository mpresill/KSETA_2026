## Detector Response

Let's simulate a simple detector response to particle energies.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate true particle energies
np.random.seed(42)
true_energies = np.random.exponential(scale=50, size=1000)

# Simulate detector resolution (Gaussian smearing)
def detector_response(true_energy, resolution=0.1):
    """
    Simulate detector response with Gaussian resolution
    resolution: fractional energy resolution (e.g., 0.1 = 10%)
    """
    sigma = resolution * true_energy
    measured_energy = np.random.normal(true_energy, sigma)
    return measured_energy

# Apply detector response
measured_energies = np.array([detector_response(E) for E in true_energies])

# Plot comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.hist(true_energies, bins=50, alpha=0.7, label='True Energy')
ax1.set_xlabel('Energy (GeV)')
ax1.set_ylabel('Counts')
ax1.set_title('True Energy Distribution')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.hist(measured_energies, bins=50, alpha=0.7, color='orange', label='Measured Energy')
ax2.set_xlabel('Energy (GeV)')
ax2.set_ylabel('Counts')
ax2.set_title('Measured Energy Distribution')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Statistical Analysis

Let's perform a simple statistical analysis of the measured data.

```python
# Calculate statistics
mean_true = np.mean(true_energies)
std_true = np.std(true_energies)
mean_measured = np.mean(measured_energies)
std_measured = np.std(measured_energies)

print(f"True Energy:")
print(f"  Mean: {mean_true:.2f} GeV")
print(f"  Std Dev: {std_true:.2f} GeV")
print(f"\nMeasured Energy:")
print(f"  Mean: {mean_measured:.2f} GeV")
print(f"  Std Dev: {std_measured:.2f} GeV")
print(f"\nRelative difference in mean: {abs(mean_measured - mean_true) / mean_true * 100:.2f}%")
```

## Systematic Uncertainties

In real experiments, we need to account for systematic uncertainties such as:
- Calibration uncertainties
- Background contamination
- Detector efficiency
- Trigger effects

```python
# Example: Energy scale uncertainty
energy_scale_uncertainty = 0.02  # 2% uncertainty

# Upper and lower variations
measured_energies_up = measured_energies * (1 + energy_scale_uncertainty)
measured_energies_down = measured_energies * (1 - energy_scale_uncertainty)

# Plot with systematic variations
plt.figure(figsize=(10, 6))
plt.hist(measured_energies, bins=50, alpha=0.5, label='Nominal', color='blue')
plt.hist(measured_energies_up, bins=50, alpha=0.3, label='Scale +2%', color='red')
plt.hist(measured_energies_down, bins=50, alpha=0.3, label='Scale -2%', color='green')
plt.xlabel('Energy (GeV)')
plt.ylabel('Counts')
plt.title('Energy Distribution with Systematic Variations')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
