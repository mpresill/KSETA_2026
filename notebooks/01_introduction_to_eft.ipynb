# Introduction to Effective Field Theory

In this notebook, we will explore the basics of Effective Field Theory (EFT).

## What is EFT?

Effective Field Theory is a framework for analyzing physical systems at different energy scales. The key idea is to separate high-energy physics from low-energy physics systematically.

### Key Concepts

1. **Separation of Scales**: We work at energies $E \ll M$ where $M$ is some high-energy scale.
2. **Power Counting**: Organize terms by their importance at low energies.
3. **Matching**: Connect the EFT to the underlying theory at the scale $M$.
4. **Running**: Evolution of parameters with energy scale.

## Example: Simple Harmonic Oscillator

Let's start with a simple example to illustrate EFT concepts using Python.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
m = 1.0  # mass
omega = 1.0  # frequency

# Time array
t = np.linspace(0, 10, 1000)

# Position for harmonic oscillator
x = np.cos(omega * t)

# Plot
plt.figure(figsize=(10, 4))
plt.plot(t, x)
plt.xlabel('Time')
plt.ylabel('Position')
plt.title('Harmonic Oscillator')
plt.grid(True)
plt.show()
```

## Dimensional Analysis

In EFT, dimensional analysis is crucial. Let's examine the dimensions of operators.

```python
# Example: Calculate mass dimension of operators
def mass_dimension(operator_type):
    """
    Returns the mass dimension of common operators in d=4 spacetime
    """
    dimensions = {
        'scalar_field': 1,
        'fermion_field': 3/2,
        'gauge_field': 1,
        'derivative': 1,
        'mass': 1,
    }
    return dimensions.get(operator_type, 'Unknown')

# Test the function
print(f"Scalar field dimension: {mass_dimension('scalar_field')}")
print(f"Fermion field dimension: {mass_dimension('fermion_field')}")
print(f"Gauge field dimension: {mass_dimension('gauge_field')}")
```
